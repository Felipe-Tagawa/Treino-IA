```python
# Regressão Linear é o método utilizado para prever valores contínuos,
# como por exemplo, o preço de uma casa, a temperatura de um dia, entre outros.
# Esse método utiliza uma equação linear (reta) para modelar a relação entre uma variável dependente e uma ou mais variáveis independentes.
# Há um erro (EQM) associado a cada previsão, que é a diferença entre o valor previsto e o valor real. 
# O objetivo da regressão linear é minimizar esse erro.
# Um resíduo negativo fica abaixo da reta.
# O primeiro ponto sempre está contido nessa reta, sendo fundamental na busca pelo coeficiente angular desta reta.

# O gradiente nablaJ é a lista de derivadas de J, uma para cada coeficiente. No mínimo, todas valem zero.
# Se a reta não apresenta a coluna ls (limite superior), o modelo será de regressão linear simples (passa pela origem).
# XtX sempre será quadrada, pois é multiplicação de linhas e colunas de uma matriz e sua transposta. 
```

```python
import numpy as np
from sklearn.datasets import fetch_california_housing

dados = fetch_california_housing()
X, y = dados.data, dados.target

print(X.shape, y.shape)
print(dados.feature_names)
```

Coluna de ls

```python
n = X.shape[0]
Xb = np.column_stack([np.ones(n), X])

print(Xb.shape)
print(Xb[0, :3])
```

Calcular B pelos três métodos

```python
A = Xb.T @ Xb
b = Xb.T @ y

beta_inv = np.linalg.inv(A) @ b
beta_solve = np.linalg.solve(A,b)
beta_pinv = np.linalg.pinv(Xb) @ y

print(np.allclose(beta_solve,beta_pinv))
print(np.abs(beta_inv - beta_pinv).max())

y_hat = Xb @ beta_pinv
print(y_hat.shape)
```

r2

```python
r2 = 1 - np.sum((y - y_hat)**2) / np.sum((y - y.mean())**2)
print(r2)
```

Conferir com Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X,y)
beta_sk = np.r_[lr.intercept_, lr.coef_]

print(np.allclose(beta_pinv, beta_sk))
print(lr.score(X,y))
```
