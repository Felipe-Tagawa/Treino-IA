```python
import numpy as np
```

```python
class GradientDescendente:
    def __init__(self, alpha=0.1, epocas=1000, tol=1e-8):
        # Guarda os hiperparâmetros
        self.alpha = alpha
        self.epocas = epocas
        self.tol = tol
        self.beta_ = None
        self.historico_ = []
    def adicionar_intercepto(self, X):
        if X.ndim == 1:
            X = X.reshape(-1, 1)
        return np.column_stack([np.ones(len(X)), X])
    
    def fit(self, X, y):
        # Adiciona 1s, treina beta_, salva historico_
        Xb = self.adicionar_intercepto(X)
        m = len(y)
        self.beta_ = np.zeros(Xb.shape[1])
        self.historico_ = []

        for t in range(self.epocas):
            y_hat = Xb @ self.beta_
            erro = y_hat - y
            custo = (erro @ erro) / (2 * m)

            self.historico_.append(custo)

            gradiente = (Xb.T @ erro) / m
            self.beta_ = self.beta_ - self.alpha * gradiente

            if t > 0 and abs(self.historico_[-2] - self.historico_[-1]) < self.tol:
                break

        return self

    def predict(self, X):
        # Retorna as previsões y

        Xb = self.adicionar_intercepto(X)
        return Xb @ self.beta_
    def score(self, X, y):
        # Retorna o R²

        y_hat = self.predict(X)
        SQres = np.sum((y - y_hat) ** 2)
        SQtot = np.sum((y - np.mean(y)) ** 2)

        return 1 - SQres/ SQtot
```

```python
# Exemplos para ver se deu certo
X_teste = np.array([[1.0], [2.0], [3.0]])
y_teste = np.array([2.0, 4.0, 6.0])

gd = GradientDescendente(alpha=0.1, epocas=1000)
gd.fit(X_teste, y_teste)

print(f"Número de épocas até convergência: {len(gd.historico_)}")
print(f"R² do modelo: {gd.score(X_teste, y_teste):.4f}")
```
