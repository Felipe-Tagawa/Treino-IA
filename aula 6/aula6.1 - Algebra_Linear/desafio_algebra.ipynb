```python
import numpy as np
import scipy as sc
```

#### Variáveis

```python
# A = np.array([[1, 2], [3, 4]])
rng = np.random.default_rng(42) # Gerador de números aleatórios
n, d = 10_000, 5
X = rng.normal(size=(n, d))
x = X[0]
w = rng.normal(size=d) # Vetor de pesos com 5 elementos
y = X @ w + rng.normal(scale=0.1, size=n) # Adicionando ruído aos rótulos
```

#### Norma L1 e L2 (segunda ordem)

```python
normaL1 = np.sum(np.abs(x)) # Somatória dos módulos
normaL2 = np.sqrt(np.sum((x ** 2))) # Raiz quadrada do somatório ao quadrado dos valores de x

print(normaL1, normaL2)

normaL1_np = np.linalg.norm(x, 1)
normaL2_np = np.linalg.norm(x, 2)

print(normaL1_np, normaL2_np)

print('BATEU' if normaL1 == normaL1_np and normaL2 == normaL2_np else 'NÃO BATEU')
```

#### Distâncias

```python
A, B = X[:200], X[:50]

diff = A[:, None, :] - B[None, :, :]

D_v = np.sqrt((diff ** 2).sum(axis=2)) # Vetorizado

D_sc = sc.spatial.distance.cdist(A, B, metric='euclidean')

print('BATEU' if np.allclose(D_v, D_sc) else 'NÃO BATEU')
```

### MSE e Gradiente na forma Matricial

```python
# MSE(X, y, w)
# GRAD_MSE(X, y, w)

w0 = np.zeros(d)
mse, grad = 0.0, np.zeros(d)

erro = X @ w0 - y # (10_000,)
mse_v = (erro ** 2).mean() # ()
grad_v = 2 / n * X.T @ erro # (5,)

def mse_de(w):
    return np.mean((X @ w - y) ** 2)

def grad_numerico(f, w, h=1e-6):
    g = np.zeros_like(w)
    for j in range(len(w)):
        e = np.zeros_like(w)
        e[j] = h
        g[j] = (f(w + e) - f(w - e)) / (2 * h)
    return g

def grad_analitico_mse(X, y, w):
    n = len(y)
    
    # Cálculo vetorizado usando o operador @ para multiplicação de matrizes
    grad = 2 / n * X.T @ (X @ w - y)
    
    return grad

print('BATEU' if np.allclose(grad_v, grad_numerico(mse_de, w0)) else 'NÃO BATEU')
```
