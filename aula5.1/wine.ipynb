# Refazendo o exercício usando o dataset wine

## Imports

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_wine
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.dummy import DummyClassifier
import matplotlib.pyplot as plt
```

## Load e Split Inicial

```python
wine = load_wine()

X = wine.data
y = wine.target

for index, class_name in enumerate(wine.target_names):
    print(f"Classe {index}: {class_name}")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2,
    stratify=y, random_state=42
)
```

## Criação do Modelo e Treinamento

```python
model = KNeighborsClassifier(
    n_neighbors=5,
)

model.fit(X_train, y_train)
```

## Predições

```python
y_pred = model.predict(X_test)
print(accuracy_score(y_test, y_pred))
print("Classes que o modelo conhece:", model.classes_)


# Gráfico de Confusion Matrix

ConfusionMatrixDisplay.from_estimator(
    model,
    X_test,
    y_test,
    display_labels=['First Region', 'Second Region', 'Third Region'], 
    cmap='Blues',
)
```

## Comparação

```python
base = DummyClassifier(
    strategy='most_frequent'
)

base.fit(X_train, y_train)

print(base.score(X_test, y_test))
```

## Resultados

O score de accuracy do modelo KNN (80,5%) foi bem maior que o do DummyClassifier;

A matriz de confusão mostra que a classe 0 foi classificada de forma integral, mas as classes 1 e 2 tiveram algumas confusões entre elas;

Como só foi especificado o valor de K-Neighbors, os outros hiperparâmetros permaneceram com o valor default para cada.
